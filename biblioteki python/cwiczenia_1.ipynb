Zaimportować funkcję `randint` z modułu `random` i nazwać ją `losuj_zakres`. Napisać program, który wykorzysta tę funkcję do wylosowania numeru studenta, który pójdzie do tablicy.

```python
import random
random.randint(5,10) #postawienie pytajnika przed funkcją pozwala na wyświetlenie jej dokumentacji
```

```python
from random import randint as losuj_zakres #importowanie jednej funkcji i nadawanie jej nazwy
num_studenta = losuj_zakres(1,11) #w ten sposób nie wystąpi błąd (do tablicy idzie i wstaje ten sam student)
print(f'do tablicy idzie student o numerze {num_studenta}')
print(f'student o numerze {num_studenta} wstaje z krzesła')
#bez przypisania randint(zakres) do jednej zmiennej to funkcja dwa razy wylosuje numery
```
