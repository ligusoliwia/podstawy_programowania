### Po co Plotly?

Matplotlib i Seaborn tworzą **statyczne** obrazki (PNG).  
Plotly tworzy **interaktywne** wykresy HTML:
- zoom / pan myszką
- hover — kliknij punkt i zobaczysz wartości
- ukrywanie serii kliknięciem w legendę
- zapis do PNG jednym kliknięciem

Żeby to działa w Jupyter-Lab:
```python
import plotly.io as pio
pio.renderers.default = "notebook"   # lub "jupyterlab"
```

---
## Podsumowanie

| Typ wykresu | Matplotlib/Seaborn | Plotly Express |
|---|---|---|
| Scatter | `plt.scatter` / `sns.scatterplot` | `px.scatter` |
| Histogram | `plt.hist` / `sns.histplot` | `px.histogram` |
| Boxplot | `plt.boxplot` / `sns.boxplot` | `px.box` |
| Violin | `sns.violinplot` | `px.violin` |
| Pairplot | `sns.pairplot` | `px.scatter_matrix` |
| Siatka wykresów | `plt.subplots` + pętla | `facet_col` / `facet_row` |
| Adnotacja punktu | `plt.annotate` | `fig.add_annotation` |

### Kiedy używać czego?

- **Matplotlib/Seaborn** — publikacje, raporty PDF, pełna kontrola nad każdym pikselem
- **Plotly** — eksploracja danych, prezentacje, dashboardy, notebooki do dalszej analizy
