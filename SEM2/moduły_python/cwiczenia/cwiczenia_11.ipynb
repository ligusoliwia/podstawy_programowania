# Ćwiczenia 11. Zaawansowane operacje z modułem NumPy

## Ćwiczenie 1. Zmiana wymiarów danych

Wygeneruj tablicę NumPy wypełnioną zerami o wymiarach $100\times 50$. 

* zmień wymiary tej tablicy na $50\times 100$
* zmień wymiary tej tablicy do $100\times 45$
* spłaszcz tą tablicę do wektora
* przywróć jej oryginalny rozmiar.

## Ćwiczenie 2. Łączenie tablic

Utwórz dwie nowe tablice wypełnione losowymi wartościami o rozmiarze 10. 
* złącz te tablice uzyskując nową tablicę o wymiarze 20
* złącz te tablice uzyskując nową tablicę o wymiarach $2\times 10$
* złącz te tablice uzyskując nową tablicę o wymiarach $10\times 2$

## Ćwiczenie 3. Łączenie tablic wielowymiarowych.

Utwórz macierz A o wymiarach $3\times 4$, macierz B o wymiarach $5\times 4$ oraz macierz C o wymiarach $3\times 5$.
* złącz macierze A i B
* złącz macierze A i C
* złącz macierze B i C
* utwórz dodatkową macierz D o wymiarach $3\times 4$ i złącz ją z macierzą A aby uzyskać tensor o wymiarach $3\times 4\times 2$

## Ćwiczenie 4. Dzielenie tablic

* podziel macierz A na kolumny
* podziel macierz B na wiersze
* podziel tensor powstały ze złączenia A i D na 3 macierze o wymiarach $4\times 2$

## Ćwiczenie 5. Broadcasting

Utwórz wektor a o rozmiarze 4. Wykonaj następujące operacje:
* pomnóż wektor a przez macierz A
* pomnóż wektor a przez macierz B
* wykonaj sumowanie wartości w macierzach A i B

## Ćwiczenie 6. Po co to wszystko?

Wygeneruj macierz E o wymiarach $154\times 12$ oraz macierz F o wymiarach $122\times 12$ obie o wartościach całkowitych z przedziału 0 do 100. Przypuśćmy, że to tablice BIOM dwóch zestawów próbek które chcemy porównać na poziomie $\beta$. Oblicz macierz odległości Euklidesa między wszystkimi parami wierszy z tych macierzy. Narysuj mapę poklastrowanej macierzy odległości.
