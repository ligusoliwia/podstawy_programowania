## Ćwiczenie 3. Łączenie tablic wielowymiarowych.

Utwórz macierz A o wymiarach $3\times 4$, macierz B o wymiarach $5\times 4$ oraz macierz C o wymiarach $3\times 5$.
* złącz macierze A i B
* złącz macierze A i C
* złącz macierze B i C
* utwórz dodatkową macierz D o wymiarach $3\times 4$ i złącz ją z macierzą A aby uzyskać tensor o wymiarach $3\times 4\times 2$

```python
macierz_A = np.empty((3,4))
macierz_B = np.empty((5,4))
macierz_C = np.empty((3,5))
```

```python
# vstack łączy przez dodanie wierszy, więc liczba kolumn musi być taka sama
# hstack łączy przez dodanie kolumn, więc liczba wierszy musi być taka sama
```

```python
np.vstack([macierz_A, macierz_B])
```

```python
np.hstack([macierz_A, macierz_C])
```

```python
np.vstack([macierz_B.T, macierz_C])
```

```python
macierz_D = np.empty((3,4))
```

```python
#tensor to tablica o więcej niż 2 wymiarach
```

```python
t = np.stack([macierz_A, macierz_D], axis=2)
```

```python
t
```

## Ćwiczenie 6. Po co to wszystko?

Wygeneruj macierz E o wymiarach $154\times 12$ oraz macierz F o wymiarach $122\times 12$ obie o wartościach całkowitych z przedziału 0 do 100. Przypuśćmy, że to tablice BIOM dwóch zestawów próbek które chcemy porównać na poziomie $\beta$. Oblicz macierz odległości Euklidesa między wszystkimi parami wierszy z tych macierzy. Narysuj mapę poklastrowanej macierzy odległości.

```python
macierz_E = np.random.randint(0, 100, size=(154, 12))
macierz_F = np.random.randint(0, 100, size=(122, 12))
```

```python
r = macierz_E[np.newaxis, :, :] - macierz_F[:, np.newaxis, :]
```

```python
k = r**2
```

```python
suma = np.sum(k, axis=2)
```

```python
red = k.sum()
```

```python
redukcja = red**0.5
```

```python
dis = suma**0.5
```

```python
redukcja
```

```python
dis
```

```python
import seaborn as sns
```

```python
sns.clustermap(dis)
```
